```python
import torch
from vllm import LLM, SamplingParams
```

```python
llm = LLM(model="../../models/hf/Meta-Llama-3.1-8B-Instruct",
          dtype=torch.float16,
          trust_remote_code=True,
          enforce_eager=True,
          max_model_len=600,
          quantization="bitsandbytes",
          load_format="bitsandbytes")
```

```python
TITLE="""Title: Benny the Bear and the Lost Honey
Benny the Bear loved nothing more than the sweet taste of honey. But one morning, when he went to check on his favorite honey tree, the honey was gone!"""

PROMPT = f"""### Write a short fairy tale of about 500 words suitable for little kids.
{TITLE}"""

sampling_params = SamplingParams(temperature=0.8,
                                 top_p=0.95,
                                 min_tokens=400,
                                 max_tokens=600,
                                 stop=['###'])

outputs = llm.generate([PROMPT], sampling_params)

story = outputs[0].outputs[0].text

print(f"{TITLE}{story}")
```

```python
PROMPT_SUMMARIZE = f"""### Summarize plot of the following story
{TITLE}{story}
### Summary
"""

sampling_params1 = SamplingParams(temperature=0.8,
                                 top_p=0.95,
                                 min_tokens=200,
                                 max_tokens=400)#, stop=['###'])

print(PROMPT_SUMMARIZE)
outputs = llm.generate([PROMPT_SUMMARIZE], sampling_params1)
summary = outputs[0].outputs[0].text
print(summary)
```
