Google's Gemma is an open-weights model you can run locally.
You can donwload the model here:

https://www.kaggle.com/models/google/gemma

It has multiple versions and multiple formats to support most of the popular
libraries.

The recommended methods of running Gemma is:

1.  Using transformers.  You can download the transformers' version and use the notebook in 05-transformers.
2.  GGUF.  You can download the GGUF and use notebook in 03-llama-cpp


Gemma has a command-line interface called "gemma.cpp".  It's obviously a `copy` of llama.cpp, but nevertheless the interface is polished and it could be fun to play with.
You have to download the gemma.cpp version of model (with filename *.sbs).

The script `build.sh` in this folder can be run under command line to download and build the gemma.cpp source code.  It will generate the executable `gemma` in the current directory.

You can see from Gemma how Google is struggling and how it is breaking down.  Their
top tier model Gemini is behind GPT and maybe even Claude.  Their open-weights
model Gemma 2 is way behind Llama 3.1. They provide API for a lot of frameworks
but model adaptation is low.  Even their own people are leaving Tensorflow to build
FLAX and JAX, and a dozen of other frameworks but people are not picking up these
frameworks either.  This is why big companies with a lot of talents cannot do things
right.  A 100 mediocre products are not going to save the company.  People do not
come to your model because you provide 10 APIs.  People learn to use the API of the winning model.
