One clean way to expose the API key to Python is to set it as an environment variable.

Add the following line to ~/.bashrc, replacing the quoted string with your API key.

export API_KEY=""

Logout and back in to make the environment variable effective.

Or you can add your API_KEY using `os.environ['API_KEY']`.

```python
import os
# 
# os.environ['API_KEY'] = ''
```

```python
import google.generativeai as genai

# You can actually directly pass in api_key instead of going through
# os.environ.  Nevertheless, directly having the key in source code is
# a very bad behavior.
genai.configure(api_key=os.environ["API_KEY"])

model = genai.GenerativeModel('gemini-1.5-flash')
```

```python
response = model.generate_content("Write a story about an AI and magic")
print(response.text)
```
