```python
# download model and change MODEL_PATH
# https://huggingface.co/chatpdflocal/llama3.1-8b-gguf/tree/main
# You can search for other llama3.1 finetunes.
MODEL_PATH = "../models/gguf/Meta-Llama-3-8B-Instruct.Q4_0.gguf"
```

```python
from llama_cpp import Llama, LlamaGrammar
import requests
import json

grammar_text = requests.get("https://raw.githubusercontent.com/ggerganov/llama.cpp/master/grammars/json_arr.gbnf").text
grammar = LlamaGrammar.from_string(grammar_text)

llm = Llama(MODEL_PATH)
```

```python
# The grammar constraint works by restricting searching to
# only generate text of valid grammar.

# For example, if we are expecting a dict, then the next
# token can only be '{'.  Whatever next token distribution
# the LLM predicts, we'll only look at those tokens that
# are grammartically acceptable.  So even if the LLM knows
# nothing about the grammar, you'll still get valid output,
# except the quality will be low.

# You need to make the LLM be aware of the grammar you want
# via prompt engineering.  If it's an uncommon grammar,
# you can provide few-shot learning samples in the prompt.

response = llm(
    "JSON list of name strings of attractions in SF:",
    grammar=grammar, max_tokens=-1
)

out = json.dumps(json.loads(response['choices'][0]['text']), indent=4)
print(out)

```
